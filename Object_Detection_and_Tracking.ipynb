```python
video_file="/mixkit-two-avenues-with-many-cars-traveling-at-night-34562-hd-ready (1).avi"
```

```python
!pip install ultralytics

from ultralytics import YOLO
import os
import glob
from google.colab import files

model = YOLO('yolov8n.pt')

print("Searching for your uploaded video...")
# This will find your video whether it is an .mp4 or your new .avi file!
uploaded_videos = glob.glob("/content/*.mp4") + glob.glob("/content/*.avi")

if len(uploaded_videos) == 0:
    print("\nSTOP: I cannot find any video files! Please upload your video and try again.")
else:
    video_file = uploaded_videos[0]
    print(f"Found it! Starting tracking on: {video_file}")

    print("Processing video frame by frame. This might take a minute...")

    save_directory = None
    for frame_result in model.track(source=video_file, tracker="bytetrack.yaml", save=True, stream=True):
        save_directory = frame_result.save_dir

    if save_directory:
        saved_videos = glob.glob(os.path.join(save_directory, "*.*"))

        if saved_videos:
            final_output_path = saved_videos[0]
            print(f"\nSuccess! Your video is ready at: {final_output_path}")
            print("Downloading to your computer now...")
            files.download(final_output_path)
        else:
            print("\nOops! Couldn't find the saved video file.")
    else:
        print("\nError: Processing failed.")
```
