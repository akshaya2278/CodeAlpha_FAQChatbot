```python
video_file="/mixkit-two-avenues-with-many-cars-traveling-at-night-34562-hd-ready.mp4"
```

```python
# --- STEP 1: Install the Ultralytics library ---
!pip install ultralytics

# --- STEP 2: Import the tools we need ---
from ultralytics import YOLO
import os
import glob
from google.colab import files

# --- STEP 3: Load the pre-trained YOLOv8 model ---
model = YOLO('yolov8n.pt')

# --- STEP 4: Define your video file ---
# The correct path with /content/ is included right here!
video_file = "/content/mixkit-two-avenues-with-many-cars-traveling-at-night-34562-hd-ready.mp4"

print(f"Starting tracking on {video_file}...")

# --- STEP 5: Run the Object Detection and Tracking ---
print("Processing video frame by frame. This might take a minute...")

save_directory = None
for frame_result in model.track(source=video_file, tracker="bytetrack.yaml", save=True, stream=True):
    save_directory = frame_result.save_dir

# --- STEP 6: Find and Download the Processed Video ---
if save_directory:
    saved_videos = glob.glob(os.path.join(save_directory, "*.*"))

    if saved_videos:
        final_output_path = saved_videos[0]
        print(f"\nSuccess! Your video is ready at: {final_output_path}")
        print("Downloading to your computer now...")

        # Trigger the browser to download the file automatically
        files.download(final_output_path)
    else:
        print("\nOops! Couldn't find the saved video file.")
else:
    print("\nError: Processing failed. Double-check your video file name.")
```
